```python
import pathlib
import re
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import xarray as xr
```

```python
SIM_ROOT = pathlib.Path("../")
sims = sorted(SIM_ROOT.glob("sim_*"))
sims
```

```python
re_dt = re.compile(r"(\d+\.\d+) elapsed seconds")
```

```python
def get_rsl_dt(f):
    rsl = f.read_text().splitlines()
    rsl = [l for l in rsl if "Timing for main" in l]
    rsl_dt = re_dt.findall("".join(rsl))
    rsl_dt = [float(t) for t in rsl_dt]
    return rsl_dt
```

```python
dt_df = {s.name: get_rsl_dt(s / "3_wrf/rsl.out.0000") for s in sims}

# Pad with NaNs to make series of equal length
max_len = max(len(v) for v in dt_df.values())
dt_df = {k: v + [float("nan")] * (max_len - len(v)) for k, v in dt_df.items()}
dt_df = pd.DataFrame(dt_df)
dt_df
```

```python
fig, ax = plt.subplots(figsize=(15, 6))
dt_df.plot(ax=ax, legend=False)
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
dt_df.plot.hist(bins=np.arange(0, 5, .05), ax=ax, cmap="tab20")
```

# Precip test in wrfout

```python
import wrf
import netCDF4
```

```python
wrfout_files = sorted((sims[0] / "3_wrf").glob("wrfout*"))
wrfout_files
```

```python
ds_list = [xr.load_dataset(f) for f in wrfout_files]
```

```python
nc_list = [netCDF4.Dataset(f) for f in wrfout_files]
```

```python
ds = ds_list[1]
ds
```

```python
ds["QRAIN"].sum(["bottom_top"]).plot(col="Time")
```

```python
wrf.getvar(nc_list[1], "dbz", timeidx=0).plot(col="bottom_top", col_wrap=10)
```

```python
Rd = 287.05
g  = 9.81

p    = ds['P'] + ds['PB']                       # full pressure, Pa, mass levels
z_w  = (ds['PH'] + ds['PHB']) / g                # geopotential height at w-levels (staggered)
dz   = z_w.diff('bottom_top_stag').rename({'bottom_top_stag': 'bottom_top'})

Tk   = wrf.getvar(nc_list[1], "tk", timeidx=None)
Tv   = Tk * (1 + 0.61 * ds['QVAPOR'])               # virtual temp
rho  = p / (Rd * Tv)                                # density, kg/m3

def column_path(q, rho, dz):
    return (q * rho * dz).sum(dim='bottom_top')


xx = column_path(ds['QRAIN'], rho, dz)
```

```python
xx.plot(col="Time")
```
